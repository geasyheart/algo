```python
from sklearn.model_selection import StratifiedKFold
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "X": list(range(100)),
    'Y': [_//10 for _ in range(100)]
})

df['Y'].value_counts()
```

```python
df
```

# 第一个特征，shuffle和不shuffle的区别

```python
skf = StratifiedKFold(n_splits=5, shuffle=False)

for fold, (train_idx, val_idx) in enumerate(skf.split(X=df['X'], y=df['Y'])):
    print(val_idx)
```

可以看到，每个类别都均匀分了1/5，例如标签0，有0、1，1有10，11
由于这里shuffle=False，所以没有打乱

```python
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=2666)

for fold, (train_idx, val_idx) in enumerate(skf.split(X=df['X'], y=df['Y'])):
    print(val_idx)
```

这里标签0就是1，6，标签1就是4，8，说明在这个label组内顺序被打乱了

# 第二个特性，所有fold val_idx集合到一起会覆盖所有样本

```python
oof = np.full(shape=df.shape[0], fill_value=-100)

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=2666)

for fold, (train_idx, val_idx) in enumerate(skf.split(X=df['X'], y=df['Y'])):
    oof[val_idx] = 1
    
    
assert (oof==1).all()
```
