```python
#假设我们测量10个产品的长度（mm）：


data = [99, 100, 101, 100, 102, 100, 101, 99, 100, 150]
```

```python
import numpy as np

# 基于3σ的方法
mu = np.mean(data)
sigma = np.std(data) + 1e-6

((mu - 3*sigma), (mu + 3*sigma))
# 可以看到150这个异常点也包含在内了
```

```python
def mad(x):
    return np.median(np.abs(x - np.median(x)))
k = 3
((np.median(data) - k * 1.4826 * mad(data)), (np.median(data) + k * 1.4826 * mad(data)))
# 可以看到150这个异常点不包含在内
```

当数据中存在少量但幅度极大的异常点时，基于均值和标准差的 3σ 方法会被异常点自身“拉宽”判定区间，从而失效；而基于中位数和 MAD 的方法由于对极端值不敏感，能够更稳定地识别离群点。
