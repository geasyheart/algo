```python
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import numpy as np

# 加载手写数字
digits = load_digits()
X_digits = digits.data
y_digits = digits.target

# 选取一个数字
sample_idx = 26
original_digit = X_digits[sample_idx].reshape(8, 8)

# 使用不同主成分数量重建
n_components_list = [2, 10, 30, 64]
reconstructed_digits = []

fig, axes = plt.subplots(1, len(n_components_list) + 1, figsize=(12, 3))

# 原始图像
axes[0].imshow(original_digit)
axes[0].set_title('原始数字')
axes[0].axis('off')

# 重建图像
for i, n_comp in enumerate(n_components_list):
    pca = PCA(n_components=n_comp)
    X_pca = pca.fit_transform(X_digits)
    X_reconstructed = pca.inverse_transform(X_pca)
    
    # 保存重建结果
    reconstructed_digit = X_reconstructed[sample_idx].reshape(8, 8)
    reconstructed_digits.append(reconstructed_digit)
    
    # 计算重建误差
    mse = np.mean((original_digit - reconstructed_digit)**2)
    
    # 显示
    axes[i+1].imshow(reconstructed_digit)
    axes[i+1].set_title(f'{n_comp}个成分\nMSE: {mse:.2f}')
    axes[i+1].axis('off')

plt.tight_layout()
plt.show()
```
